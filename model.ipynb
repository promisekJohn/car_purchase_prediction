```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("car.csv")
df
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
le = LabelEncoder()
le
```

```python
# Encode the Categorical Columns
df["Gender"] =le.fit_transform(df["Gender"])
df
```

```python
# Drop CustomerID
df = df.drop("CustomerID", axis=1)
```

```python
# Select Features (x)
x = df.drop("PurchaseCar", axis=1)
x
```

```python
y = df["PurchaseCar"]
y
```

```python
from sklearn.model_selection  import train_test_split
```

```python
# Split the Dataset
x_train, x_test, y_train, y_test = train_test_split(x,y,test_size=0.2,random_state=42)
```

## Logistics Regression

```python
from sklearn.linear_model import LogisticRegression
```

```python
TLRM = LogisticRegression()  
```

```python
TLRM.fit(x_train, y_train)
```

```python
PLRM =  TLRM.predict(x_test)
```

```python
PLRM
```

## Decision Tree

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
DTM = DecisionTreeClassifier()
```

```python
DTM.fit(x_train, y_train)
```

```python
PDTM = DTM.predict(x_test)
```

```python
PDTM
```

## Random Forest

```python
# importing the library
from sklearn.ensemble import RandomForestClassifier
```

```python
RFCT = RandomForestClassifier()
```

```python
RFCT.fit(x_train, y_train)
```

```python
RFCP = RFCT.predict(x_test)
RFCP
```

## Acuracy Score

```python
from sklearn.metrics import accuracy_score
```

```python
## ACCURACY SCORE FOR LOGISTIC MODEL
accuracy_score(y_test, PLRM) * 100
```

```python
## ACCURACY SCORE FOR DECISION TREE MODEL
accuracy_score(y_test, PDTM) * 100
```

```python
## ACCURACY SCORE FOR RANDOM FOREST MODEL
accuracy_score(y_test, RFCP) * 100
```

## confusion matrix

```python
from sklearn.metrics import confusion_matrix
```

```python
# CONFUSION MATRIX FOR LOGISTICS MODEL
confusion_matrix(y_test, PLRM)
```

```python
# CONFUSION MATRIX FOR DECISION MODEL
confusion_matrix(y_test,PDTM)
```

```python
# CONFUSION MATRIX FOR RANDOM FOREST MODEL
confusion_matrix(y_test,RFCP)
```

### CLASSIFICATION REPORT|

```python
from sklearn.metrics import classification_report
```

```python
# CLASSIFICATION REPORT FOR LOGISTICS MODEL
print(classification_report(y_test, PLRM))
```

```python
# CLASSIFICATION REPORT FOR DECISION MODEL
print(classification_report(y_test,PDTM))
```

```python
# CLASSIFICATION REPORT  FOR RANDOM FOREST MODEL
print(classification_report(y_test, RFCP))
```
